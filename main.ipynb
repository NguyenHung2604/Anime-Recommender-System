```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
# Phan tich muc do tuong quan giua cac feature

# Du lieu so
df_anime = pd.read_csv(anime_path)

# Chuyen doi cac tap bi unknown tro ve NAN
df_anime['episodes'] = pd.to_numeric(df_anime['episodes'], errors='coerce')

# numerical column
numerical_df = df_anime[['episodes', 'rating', 'members']]

# ma tran tuong quan
corr_matrix = numerical_df.corr()
# ve bieu do nhiet
plt.figure(figsize=(6, 8)) # kich thuoc bieu do
sns.heatmap(corr_matrix, annot = True, cmap= 'RdYlBu', fmt= '.2f', linewidths=0.5, vmin=-1, vmax= 1)
plt.title("Ma trận tương quan giữa Số tập, Điểm số và Số lượng thành viên")
plt.show()
```

```python
# kiem tra type anh huong den diem so nhu the nao 
plt.figure(figsize=(10, 6))

# Vẽ biểu đồ Boxplot: Trục X là Type, Trục Y là Rating
sns.boxplot(x='type', y='rating', data=df_anime, palette='Set3')

plt.title('Phân bố điểm số (Rating) theo từng loại hình Anime (Type)')
plt.xlabel('Loại hình (Type)')
plt.ylabel('Điểm số (Rating)')
plt.show()
```

```python
# kiem tra genre anh huong den rating nhu the nao
df_clean = df_anime.dropna(subset=['genre', 'rating']).copy()

df_clean['genre'] = df_clean['genre'].str.split(', ')

# phan ra tung the loai mot ra de xem the loai do co bao nhieu phim
df_explode = df_clean.explode('genre')

# tinh toan diem trung binh tren mot the loai anime
genre_stat = df_explode.groupby('genre').agg(
    average_rating = ('rating', 'mean'), # dung phuong thuc mean de tinh rating
    anime_count = ('anime_id', 'count') # dung count de tinh so luong anime
).reset_index()

# loc, sap xep
valid_genres = genre_stat[genre_stat['anime_count'] >= 50] # loc bo cacthe loai phim it nhung diem cao

# sap xep giam dan
top_genres = valid_genres.sort_values(by='average_rating', ascending=False)

# in ra head 10
print(top_genres.head(10))

# truc quan hoa
plt.figure(figsize=(12,6))
sns.barplot(x = 'average_rating', y = 'genre', data = top_genres, palette = 'magma')
plt.title('Top 10 Thể loại Anime có điểm đánh giá trung bình cao nhất', fontsize=14)
plt.xlabel('Điểm số trung bình (Rating)', fontsize=12)
plt.ylabel('Thể loại (Genre)', fontsize=12)

# Thu hẹp trục X (ví dụ từ điểm 6 đến 8.5) để dễ dàng nhìn thấy sự chênh lệch
plt.xlim(0, 8.5) 
plt.tight_layout()
plt.show()
```
